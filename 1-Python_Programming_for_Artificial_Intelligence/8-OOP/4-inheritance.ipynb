```python
# parent - child or    animal -- monkey - bird
```

```python
#parent

class Animal:
    def __init__(self):
        print("animal is created")

    def toString(self):
        print("animal")

    def walk(self):
        print("animal walk")
        
        
```

```python
#child
class Monkey(Animal):  # object yazsam alamazdım walk metodunu mesela 
    def __init__(self):
        super().__init__() # use init of parent(animal) class
        print("monkey is created")

    def toString(self):
          print("monkey")

    def climb(self):
        print("monkey can climb")
    # def walk eklemeye gerek yok çünkü onu class animal içinden alıcaz ortak özellik




class Bird(Animal):  # bird monkeye de bağlanabilirdi o zaman mesela climb metodu gelirdi vb. 
    def __init__(self):
        super().__init__()
        print("bird is created")

    def fly(self):
        print("fly")
```

```python
m1 = Monkey()
```

```python
m1.toString()
```

```python
m1.walk()  # parenttan aldı mesela
```

```python
m1.climb()
```

```python
b1=Bird()
```

```python
b1.walk()
```

```python
b1.climb()
```

```python
b1.fly()
```

```python
# parent --website  child id , email 
```

```python
class Website:
    "parent"
    def __init__(self, name, surname):
        self.name = name
        self.surname = surname
    
    def loginInfo(self):
        print(self.name + " "+ self.surname )
```

```python
class Website1(Website):
    "child"
    def __init__(self, name, surname, ids):
        Website.__init__(self, name, surname)
        self.ids = ids
    
    def login(self):
        print(self.name + " "+ self.surname + " "+self.ids)
```

```python
class Website2(Website):

    def __init__ (self, name, surname, email):
        Website.__init__(self,name,surname)
        self.email = email

    def login(self):
        print(self.name + " "+ self.surname + " "+self.email)
```

```python
p1 = Website("name","surname")
p2 = Website1("name","surname", "123")
p3 = Website2("name","surname", "email@")
```

```python
p1.name()
```

```python
p1.loginInfo()
```
