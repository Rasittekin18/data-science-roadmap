```python
# Kapsülleme : metot ya da variable erişimi kısıtlamak anlamına gelmektedir
# Herhangi bir nesnenin metotlarını, verilerini ve değişkenlerini diğer nesnelerden saklayarak ve bnlara erişimi sınırlandırarak
# yanlış kullanımdan koruma konseptidir.
```

```python
class BankAccount(object):

    def __init__(self,name,money,address):
        self.name=name # global
        self.__money=money  # private
        self.address=address

    # get and set  # erişebilmek için bu fonksiyon yazılır
    
    def getMoney(self):   # paraya ulaşmak için kullanılır.
        return self.__money
        
    def setMoney(self, amount):  # para çekme ya da yatırma
        self.__money = amount

     # private
    def __increase(self):
        self.__money = self.__money + 500


p1 = BankAccount("messi",1000,"barcelona")
p2 = BankAccount("neymaar",2000, "paris" )
```
