```python
class Animal: # parent
    
    def toString(self):
        print("animal")
        
class Monkey(Animal):
    
    def toString(self):
        print("monkey")
        
```

```python
a1 = Animal()
a1.toString()
```

```python
m1 = Monkey()
m1.toString() # monkey calls overriding method
```

```python
# kendi classına ait metodu cagırır . Monkeyin to stirng metodu anımalın to strıng metodunu overrıde edıyor yani gecersiz kılıyor.
```
