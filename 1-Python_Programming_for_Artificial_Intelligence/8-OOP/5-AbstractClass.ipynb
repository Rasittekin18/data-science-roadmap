```python
from abc import ABC, abstractmethod
class Animal(ABC) : #super class
    @abstractmethod   # bu şekilde animal class abstract oldu
    def walk(self): pass
    @abstractmethod  # bu şekilde yapınca mesela hem walk hem run sub da kllanılmalı yoksa hata verir.
    def run(self): pass
```

```python
class Bird(Animal):  #sub  class
    def __init__(self):
        print("bird")

    def walk(self):
        print("walk")

    def run(self):
        print("run")
```
