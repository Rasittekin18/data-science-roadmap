```python
class Emp(object):
    
    age = 25 #
    salary = 1000 # $
    
    def ageSalaryRatio(self):
        a = self.age / self.salary
        print("method: ", a)
    
e1 = Emp()
e1.ageSalaryRatio()
```

```python
# function
def ageSalaryRatio(age, salary):
    a = age / salary
    print("function: ",a)
    
ageSalaryRatio(30, 3000)
```

```python
def findArea(a, b):  # a = pi,  b = r 
    area = a*b**2
    # print(area)
    return area
    
pi = 3.14
r = 5

result1 = findArea(pi, r)
print(result1)
result2 = findArea(pi, 10)
print(result2)
print(result1 + result2)
```

```python
class Animal(object):
    
    def __init__(self, a, b): # ( name, age) = ("dog", 2) = (a, b)
        self.name = a  #self sebebi dışaardaki ile içerdeki tanınsın diyedir.
        self.age = b
    
    def getAge(self):
        print("")
        return self.age
    
    def getName(self):
        print(self.name)
```

```python
a1 = Animal("dog", 2)
a2 = Animal("cat",4)
a3 = Animal("bird", 6)
```
