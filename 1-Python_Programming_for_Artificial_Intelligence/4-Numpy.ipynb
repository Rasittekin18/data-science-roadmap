```python
# %% stacking arrays

array1 = np.array([[1,2],[3,4]])
array2 = np.array([[-1,-2],[-3,-4]])
print(array1)
print(array2)
```

```python
# veritical
#array([[1, 2],
#       [3, 4]])
#array([[-1, -2],
#       [-3, -4]])
array3 = np.vstack((array1,array2))
array3
```

```python
# horizontal
#array([[1, 2],[-1, -2],
#       [3, 4]],[-3, -4]]

array4 = np.hstack((array1,array2))
array4
```

```python
#Soru

array1= np.array([[1,2],[3,4]])
array2= np.array([[-1,-2],[-3,-4]])
np.vstack((array1,array2))
```
