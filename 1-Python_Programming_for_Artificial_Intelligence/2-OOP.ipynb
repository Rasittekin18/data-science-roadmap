## 1-OOP


```python
class Calisan:
   #self dışardan gelen ismi çalışan  classı içinde kullanmamızı sağlar.

    zam_orani=1.8 #class variable
    counter=0    # kaç çalışan olduğunu bulmak için kullanılır. 
    def __init__(self, isim, soyisim, maas):  # constructor
        self.isim = isim
        self.soyisim = soyisim
        self.maas = maas
        self.email = isim + soyisim + "@abc.com"
        #self.counter= self.counter +  1 #  obje  bu kısımda lş. için burda counter oluşturuldu.
        Calisan.counter= Calisan.counter+1
    def giveNameSurname(self):
        return self.isim + " " + self.soyisim

    def zam_yap(self):
        self.maas =  self.maas  +  self.maas* self.zam_orani #class variableına erişim  için self yazdık
isci_1 = Calisan("rasit", "tekin", 100)

    
```
